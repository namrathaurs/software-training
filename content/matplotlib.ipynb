## **The Figure-Axes Model**

The Figure-Axes model is the core concept that can make or break your plotting efforts when working with Matplotlib. 

Imagine you have a blank canvas and you are drawing something on that canvas. 

The <b>Figure</b> represents the entire canvas. The <b>Axes</b> represents the actual plotting area.

```
Figure
+----------------------------------+
|                                  |
|   Axes                           |
|   +--------------------------+   |
|   |                          |   |
|   |        Plot Here         |   |
|   |                          |   |
|   +--------------------------+   |
|                                  |
+----------------------------------+
```

## **Putting it all together...**

````python
import matplotlib.pyplot as plt

# data to be visualized goes here...
x = [1, 2, 3, 4, 5]
y = [50, 60, 70, 80, 90]

# creating the figure and axes here...
fig, ax = plt.subplots()

# creating the plot itself here...
ax.plot(x, y)

# adding labels to your plot here...
ax.set_title()
ax.set_xlabel()
ax.set_ylabel()

# save the current active figure to a file on your local disk (e.g., as PNG, JPG, PDF, or SVG)
plt.savefig("figure.png", dpi=300)

# display the figure and render your graph here afterward...
plt.show()
````
