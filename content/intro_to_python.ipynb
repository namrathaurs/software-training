v_y(t) = v_{y0} - gt.

- $v_0$ is the initial speed,
- $\theta$ is the launch angle,
- $y_0$ is the initial height,
- $g$ is the acceleration due to gravity,
- $t$ is the time after launch.

#### **`append()` vs. `extend`**

**append(item):**
* Adds exactly one element to the list.
* Increases the length of the list by 1 regardless of whether the item is a number, a string, or another list.
* If you append a list, it becomes a nested list inside the original.

**extend(iterable):**
* Unpacks the elements of an iterable (like a list, tuple, or string) and adds them one by one.
* Increases the list length by the number of elements in the iterable.
* Effectively merges or concatenates another list into the current one

````python
my_list = [1, 2, 3]
my_list.append([4, 5])
print(my_list)
# Output: [1, 2, 3, [4, 5]]

my_list = [1, 2, 3]
my_list.extend([4, 5])
print(my_list)
# Output: [1, 2, 3, 4, 5]
````

## **Tuples**

A tuple is a Python data structure that can store multiple values.

A tuple is similar to a list, but there is one important difference:

- a **list** can be changed after it is created,
- a **tuple** cannot be changed after it is created.

This means that tuples are **immutable**.

| Characteristic               | Description   |
|------------------------------|---------------|
| **Ordered**                  |     yes       |
| **Mutable**                  |     no        |
| **Allows duplicate values**  |     yes       |

Tuples are useful when we want to group values together but do not want those values to accidentally change.

### **Creating Tuples**

````python
# Example: Creating a tuple of student information
student = ("Alice", 25, 9.7)

print(student)
````

````python
# Creating an empty tuple
emptyTuple = ()
````

Since tuples are ordered, we can access their values using indices.

````python
student = ("Alice", 25, 9.7)

print(student[0])
print(student[1])
print(student[2])
````

Unlike lists, tuples cannot be modified after they are created.

````python
student = ("Alice", 25, 9.7)

# This will give an error because tuples are immutable
student[0] = "Bob"
````

If we really need to modify a tuple, one option is to convert it into a list first.

````python
student = ("Alice", 25, 9.7)

# Convert the tuple into a list
student_list = list(student)

# Modify the list
student_list.append("Physics")

# Convert the list back into a tuple
student = tuple(student_list)

print(student)
````

When it comes to tuples, **packing** means putting several values into one tuple.

````python
student = ("Alice", 25, 9.7)
````

And **unpacking** means taking the values out of the tuple and storing them in separate variables.

````python
student = ("Alice", 25, 9.7)

name, age, grade = student

print(name)
print(age)
print(grade)
````

Tuples are especially useful when a block of code (such as a function) returns more than one value.

````python
point = (3.0, 4.0)

x, y = point

print("x-position:", x)
print("y-position:", y)
````

### **Other cool operations**

The union of two sets gives us all values that appear in either set.

````python
inner_planets = {"Mercury", "Venus", "Earth", "Mars"}
visited_bodies = {"Earth", "Moon", "Mars"}

all_labels = inner_planets.union(visited_bodies)

print(all_labels)
````

The intersection of two sets gives us only the values that appear in both sets.

````python
inner_planets = {"Mercury", "Venus", "Earth", "Mars"}
visited_bodies = {"Earth", "Moon", "Mars"}

common_labels = inner_planets.intersection(visited_bodies)

print(common_labels)
````

In the world of programming, a control structure is a general construct that dictates the overall flow of execution in a program. A control structure is any block of code that manages the order in which statements are executed. It moves beyond simple line-by-line execution to allow for logic and repetition.

This is a specific subset of control structures used for decision-making and selects different execution paths based on the defined criteria. It uses boolean expressions to determine which path the program should follow.

| `else`    | Executes a block of code if none of the preceding conditions are true. |

### **if:**

Executes a block of code only if the condition evaluates to true.

````python
# Example: Checking if a number is positive.

num = int(input("Enter a number: "))
if num > 0:
    # if condition is evaluated to true, this code block will be executed.
    print("The number is positive.")
# if the condition is evaluated as false, the code block associated with if will be skipped entirely.
print("The number is not positive.")
````

### **if-else:**

Provides an alternative path when the `if` condition is false.

````python
# Example: Checking if a person is eligible to vote.

age = int(input("Enter the age of the person: "))
if age >= 18:
    # if condition is evaluated to true, the following code block will be executed.
    print("You are eligible to vote.")
else:
    # if condition is evaluated to false, the below code block will be executed, thus
    # ensuring one of two mutually exclusive actions always occurs
    print("You are not eligible to vote.")
````

### **if-elif-else:**

Used to check multiple conditions sequentially until one is found to be true.

````python
# Example: Classifying a number.

number = int(input("Enter a number: "))
# Python checks each condition in order; as soon as one matches, the corresponding block of code runs and the rest are skipped
if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    # the else block serves as a final "catch-all" if no previous conditions are met.
    print("Zero")
````

### **for:**

When we need to go through (**iterate** over) a collection of elements, we can use a `for` loop. For example, if we have a list of students, we can use a for loop to print their names and ages to the console. <u>This kind of loop is best suited when you know the number of repetitions (**iterations**) in advance or want to process every item in a collection.</u>

### **while:**

A `while` loop is used to repeatedly (**iteratively**) execute a block of code multiple times as long as a specified condition remains `True`. The while loop takes a condition just like an `if` statement. The difference is that the code block is executed multiple times until the condition is `False`. <u>This kind of loop is most useful when the number of iterations is not known beforehand.</u>

````python
# Example 1: Standard Counter

count = 1
while count <= 5:
    print(count)
    count += 1
````

````python
# Example 2: Indefinite Iteration (User Input)

user_input = ""
while user_input.lower() != "quit":
    user_input = input("Enter something (type 'quit' to exit): ")

# Description: This loop continues to prompt the user until they type "quit".
# Since we don't know how many times the user will enter data, a Python while loop is the correct tool.
````

### **`break` statement**

This control flow tool is used to exit from a loop, regardless of whether the loop's original condition is still true or if there are more iterms to iterate through. When the `break` statement is encountered inside a `for` or `while` loop, the loop gets immediately terminated and the control transfers to the next line of code after the loop.

````python
# Example 1: Breaking out of a `for` loop

fruits = ["apple", "banana", "cherry", "date"]
for fruit in fruits:
    if fruit == "cherry":
        break
    print(fruit)
````

````python
# Example 2: Breaking out of a `while loop`

while True:
    user_input = input("Type 'exit' to stop: ")
    if user_input == 'exit':
        break
    print(f"You typed: {user_input}")

# Description: The `while True` loop would run forever unless the break statement is triggered.
````

## **Classes and Objects**

**Classes** provide a means of bundling data and functionality together. So far, we have used a combination of variables, lists, dictionaries, and functions to represent data such as students. We can make our code even more organized by representing students as single *objects*. This is where classes come in.

Creating a new class creates a new type of **object**, allowing new instances of that type to be made. Each class instance can have <u>**properties**</u> (variables) attached to it for maintaining its state. Additionally, each class instance can also have <u>**methods**</u> (functions) for modifying its state.

You can think of class as a blueprint for creating something and an object is an actual thing built from that blueprint. Almost everything in Python is an object, with its variables and functions.

Imagine you are building a game with many different animals. Instead of creating new variables for every single animal, you create one blueprint that defines what an animal is.

````python
# Example: Defining a class to represent an animal

class Animal:

    # This is the constructor for this class
    def __init__(self, name, age):
        # Below (LHS) are the properties in this class
        self.name = name
        self.age = age

    # This is a method in this class
    def talk(self):
        print("Rarrrr")
````

The `Animal` class above defines the blueprint for an animal. Couple of keywords we encountered above:
* **\_\_init\_\_:** Refers to the constructor, which is a special method that runs automatically when you create a new object.

* **self:** Refers to the specific object you are working with

To create an actual animal, we need to create an object of the `Animal` class. This object is called an **<u>instance</u>** of the class.

````python
dog = Animal("Buddy", 3)
cat = Animal("Garfield", 4)
horse = Animal("Grace", 2)
lion = Animal("Simba", 7)
````

Once we have actual animal objects, we can interact with their properties and methods.

````python
print(f"{dog.name}, the dog, is {dog.age} years old.")
print(f"{dog.name} says:")
dog.talk()
print()
print(f"{cat.name}, the cat, is {cat.age} years old")
print(f"{cat.name} says:")
cat.talk()
print()
print(f"{horse.name}, the horse, is {horse.age} years old")
print(f"{horse.name} says:")
horse.talk()
print()
print(f"{lion.name}, the lion, is {lion.age} years old")
print(f"{lion.name} says:")
lion.talk()
````

## **Inheritance**

We can now use the Animal class to create a variety of animals. But what if we want to be more specific?

Let's take Buddy, for example. Buddy is a dog, but we only know that because we know Buddy. The Animal class is not specific enough to know that Buddy is a dog and Max is a cat. This is where inheritance comes in.

Inheritance allows us to define a class that inherits all the properties and methods from another class. The class that inherits is called the **child class**, and the class that is inherited from is called the **parent class**.

To understand inheritance, let us create a couple more classes.

````python
# creating a Dog class that inherits from Animal...
class Dog(Animal):
    species = "dog"

# creating a Cat class that also inherits from Animal...
class Cat(Animal):
    species = "cat"

dog = Dog("Buddy", 3)
cat = Cat("Max", 4)

print(f"{dog.name} is a {Dog.species}.")
print(f"{cat.name} is a {Cat.species}.")
print(f"{dog.name} says:")
dog.talk()
print(f"{cat.name} says:")
cat.talk()
````

Did you notice that we did not have to define the `__init__()` function in the `Dog` class? That is because the `Dog` class inherits the `__init__()` function from the `Animal` class. The same is true for the other methods and properties.

## **Polymorphism**

We still have a problem with our animals. Isn't it a bit weird that all animals make a sound in the same way? Cats don't bark, and dogs don't meow. This is where polymorphism comes in.

A child class can also override the methods and properties of its parent class to make them more specific. This is called polymorphism.

````python
# Let's redefine the Dog and Cat classes

class Dog(Animal):
    species = "dog"

    # Overriding the talk method (polymorphism)
    def talk(self):
        print("Woof")


class Cat(Animal):
    species = "cat"

    # Overriding the talk method (polymorphism)
    def talk(self):
        print("Meow")


dog = Dog("Buddy", 3)
cat = Cat("Max", 4)

print(f"{dog.name} is a {Dog.species}")
print(f"{cat.name} is a {Cat.species}")
print(f"{dog.name} says:")
dog.talk()
print(f"{cat.name} says:")
cat.talk()
````

## **Hands-on Exercise: <u>OOP</u>**



Let's practice object-oriented programming with a coding exercise that involves creating a class to represent students and interacting with the objects of the class. Here are your requirements to design and develop this class:

1. Create a class called `Student` to serve as a blueprint for storing information about students.
2. The class should have five properties: `name`, `student_id`, `major`, `courses` and `grades`.
   * `courses` should be of `list` type and must be initialized to an empty list in the class.
   * `grades` should be of `dictionary` type and must be initialized to an empty dictionary in the class.
3. The class should have five methods: `get_info`, `enroll_course`, `add_grade`, and `get_gpa`.
   * `get_info()` will be used to display student information. It should print a message in the following format and should not return any value:
      ```
      Student: John Doe (ID: 12345) - Major: Physics
      ```
   * `study()` will be used to display information about what the student is studying. The method should not return any value and print a message in the format:
      ```
      The student is studying general subjects.
      ```
   * `enroll_course()` will be used to add a course to the student's schedule. The method should:
      * accept one parameter named `course_code` and
      * add the course code to `courses` if the course code is not found in `courses` property of the object.
   * `add_grade()` will be used to record a grade for a specific course. This method should:
      * accept two parameters named `course_code` and `grade` and
      * assign the grade corresponding to the course code in `grades`.
   * `get_gpa()` will be used calculate and return the GPA of the student. This method should:
      * accept no parameters and
      * return a value of zero if `grades` is empty and
      * return the average of the grades in `grades`
4. Each of the five methods should have a multi-line style comment with a one-line description of what the method does.

````python
class Student:
    def __init__(self, name, student_id, major):
        # Initialize basic attributes
        self.name = name
        self.student_id = student_id
        self.major = major
        self.courses = []
        self.grades = {}

    def get_info(self):
        """Display student information."""
        print(f"Student: {self.name} (ID: {self.student_id}) - Major: {self.major}")
        return

    def study(self):
        """Display study information."""
        print(f"The student is studying general subjects.")
        return

    def enroll_course(self, course_code):
        """Add a course to the student's schedule."""
        if course_code not in self.courses:
            self.courses.append(course_code)

    def add_grade(self, course_code: str, grade: float):
        """Record a grade for a specific course."""
        self.grades[course_code] = grade

    def get_gpa(self) -> float:
        """Calculate and return the average GPA."""
        if not self.grades:
            return 0.0
        return sum(self.grades.values()) / len(self.grades)
````

### **File modes**

There are four modes that can be used when handling files using Python. The four modes are as follows:

**Read Mode (`r`):** To read from a file. E.g. reading a lab report data file for getting the raw data.

**Write Mode (`w`):** To write to a file. E.g. writing a simulation output file (e.g., a .txt or .csv file) to store results of experiments.

**Append Mode (`a`):** To append to the end of a file. E.g. writing data to a simulation file, adding a new sensor reading to the results.

**Read & Write Mode (`r+`):** To read from and write to a file. E.g. writing the results of a physics simulation to a data file for later analysis.
