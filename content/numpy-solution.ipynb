# **NumPy**



NumPy is the foundation of scientific computing in Python. It is an open-source Python library that is used for scientific computing, data science and machine learning. NumPy lets us represent, manipulate, and analyze large datasets efficiently using operations that closely resemble mathematical notation. 

Whether you're studying physics, engineering, biology, chemistry, or data science, NumPy is often the first computational tool you'll reach for. It allows scientists and engineers to work with entire datasets at once.

Imagine you've collected 100,000 temperature measurements, voltages, distances, or sensor readings. You want to:
* Add a calibration offset

NumPy gives us tools to do this efficiently.

## **Operations**

Instead of writing code to loop through each number in the array and perform an operation, NumPy lets us operate on the entire array (your dataset). 

````python
# adding a calibration offset to the recorded temperatures...

temperatures + 0.5
````

````python
# convert units (Celsius to Kelvin)

temperatures + 273.15
````

````python
# you can also scale values in the array...

temperatures * 2
````

````python
# also, square values...

temperatures ** 2
````

The code examples above demonstrate that one line of code involving NumPy arrays can process thousands or millions of numbers (experimental measurements, simulation outputs, images, among others).

## **Scientific Calculations**

Performing fundamental statistical calculations on the data is ubiquitous in every STEM field. Given a NumPy arrary representing measurements:

````python
measurements = np.array([8, 10, 12, 14, 16])
````

````python
# calculating mean is as simple as...

numpy.mean(measurements)
````

````python
# calculating standard deviation is as simple as...

numpy.std(measurements)
````

````python
# calculating the maximum quantity is as simple as...

numpy.max(measurements)
````

````python
# calculating the minimum quantity is as simple as...

numpy.min(measurements)
````

## **Finding Interesting Data**

You can also use NumPy arrays to identify numbers of interest. Think of identifying:

* Temperatures above a threshold
* High-energy events
* Large earthquakes
* ...

````python
# assume our numpy array
energy = numpy.array([5, 12, 18, 4, 22, 15])
````

````python
# finding values above 10
energy > 10
# the above creates a mask!
````

````python
# use the created mask on the numpy array for selection or filtering
energy[energy > 10]
````
